```python
from PyPDF2 import PdfReader, PdfWriter
```

```python
pdf_files = ["ASSIGNMENT 2.pdf", "Shashwat_Upadhyay.pdf.pdf", "3rd COMPUTER SCIENCE & IT.pdf"]
```

```python
writer = PdfWriter()
```

```python
for pdf in pdf_files:
    with open(pdf, "rb") as f:
        reader = PdfReader(f)
        for page in reader.pages:
            writer.add_page(page)
```
